```python
import pandas as pd
from sqlalchemy import create_engine

DB_USER = "root"
DB_PASSWORD = "root"
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "commerceiq_db"
```

```python
DATABASE_URL = f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"

engine = create_engine(DATABASE_URL)
```

```python
# ==========================================
# CUSTOMERS TABLE
# ==========================================

customers_df = pd.read_csv(
    "../data/raw/olist_customers_dataset.csv"
)

customers_df = customers_df[
    [
        "customer_id",
        "customer_unique_id",
        "customer_city",
        "customer_state"
    ]
]

print(customers_df.head())

customers_df.to_sql(
    name="customers",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# PRODUCTS TABLE
# ==========================================

products_df = pd.read_csv(
    "../data/raw/olist_products_dataset.csv"
)

products_df = products_df[
    [
        "product_id",
        "product_category_name"
    ]
]

print(products_df.head())

products_df.to_sql(
    name="products",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# SELLERS TABLE
# ==========================================

sellers_df = pd.read_csv(
    "../data/raw/olist_sellers_dataset.csv"
)

sellers_df = sellers_df[
    [
        "seller_id",
        "seller_city",
        "seller_state"
    ]
]

print(sellers_df.head())

sellers_df.to_sql(
    name="sellers",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# ORDERS TABLE
# ==========================================

orders_df = pd.read_csv(
    "../data/raw/olist_orders_dataset.csv"
)

orders_df = orders_df[
    [
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date"
    ]
]

print(orders_df.head())

orders_df.to_sql(
    name="orders",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# PAYMENTS TABLE
# ==========================================

payments_df = pd.read_csv(
    "../data/raw/olist_order_payments_dataset.csv"
)

payments_df = payments_df[
    [
        "order_id",
        "payment_type",
        "payment_value"
    ]
]

print(payments_df.head())

payments_df.to_sql(
    name="payments",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# REVIEWS TABLE
# ==========================================

reviews_df = pd.read_csv(
    "../data/raw/olist_order_reviews_dataset.csv"
)

reviews_df = reviews_df[
    [
        "review_id",
        "order_id",
        "review_score"
    ]
]

print(reviews_df.head())

reviews_df.to_sql(
    name="reviews",
    con=engine,
    if_exists="append",
    index=False
)


# ==========================================
# ORDER ITEMS TABLE
# ==========================================

order_items_df = pd.read_csv(
    "../data/raw/olist_order_items_dataset.csv"
)

order_items_df = order_items_df[
    [
        "order_id",
        "product_id",
        "seller_id",
        "price",
        "freight_value"
    ]
]

print(order_items_df.head())

order_items_df.to_sql(
    name="order_items",
    con=engine,
    if_exists="append",
    index=False
)
```
